```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
def npy_to_dataframe(npy_path, train_csv_path, reshape_dims=(85*17, 10)):
    """
    Converte um arquivo .npy sintético para um DataFrame Pandas,
    aplicando o escalonamento reverso baseado no dataset real.
    """
    # 1. Carregar o arquivo .npy
    synth_data = np.load(npy_path)
    
    if (len(synth_data.shape)>2):
        # 2. Reshape e conversão para DataFrame
        synth_df = pd.DataFrame(synth_data.reshape(reshape_dims))
    else:
        synth_df = pd.DataFrame(synth_data)
    
    # 3. Carregar o dataset real para obter o escalonador
    real_df = pd.read_csv(train_csv_path)
    
    # 4. Ajustar o MinMaxScaler nos dados reais
    scaler = MinMaxScaler().fit(real_df)
    
    # 5. Aplicar o inverso do escalonamento (Inverse Transform)
    synth_scaled_data = scaler.inverse_transform(synth_df)
    
    # 6. Retornar como DataFrame com os nomes das colunas originais
    return pd.DataFrame(synth_scaled_data, columns=real_df.columns)
```
